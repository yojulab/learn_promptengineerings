##실습1.6. OpenAI API 호출 및 테스트

```python
# 모델에게 보낼 메시지 구성 (역할: user)
message = {"role": "user", "content": "안녕하세요, 오늘 날씨는 어떤가요?"}

# Chat Completions API 호출 (gpt-4o-mini 모델 사용 예)
response = client.chat.completions.create(
    model="gpt-4o-mini",
    # model="gpt-5-mini",
    messages=[ message ]
)

# 응답에서 assistant의 답변 내용 추출
assistant_reply = response.choices[0].message.content
print("ChatGPT 답변:", assistant_reply)
```

## 출력 형식 제어 실습 : json_object, 코드 블럭이 생기지 않음

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "오늘 서울 날씨를 JSON 형태로 제공해줘. 키는 weather와 temperature로 설정해줘."}
    ],
    response_format={"type": "json_object"}
)

print(response.choices[0].message.content)
# 출력: '{"weather": "맑음", "temperature": "22도"}'
```
