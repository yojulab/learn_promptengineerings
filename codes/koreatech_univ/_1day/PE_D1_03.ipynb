```python
from google.colab import userdata
import os

# Colab에 저장한 Secret에서 API 키 가져와 Linux 가상 환경 변수에 등록
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
```

```python
import openai
client = openai.OpenAI()
```

```python
prompt = "서울과 부산 두 도시의 현재 날씨(온도, 날씨 상태)를 JSON 배열로 알려주세요."
```

```python
response = client.chat.completions.create(
    messages=[
        {"role": "user", "content": prompt}
    ],
    model="gpt-4o-mini")
```

```python
content = response.choices[0].message.content
print(content)  # 우선 출력 내용을 확인해봅니다.
```

```python
response = client.chat.completions.create(
    messages=[
        {"role": "user", "content": prompt}
    ],
    model="gpt-4o-mini",
    response_format={
    "type": "json_object"
    })
```

```python
content = response.choices[0].message.content
print(content)  # 우선 출력 내용을 확인해봅니다.
```

```python
import json
data = json.loads(content)
```

```python
print(data)
print(type(data))
```

```python
data.keys()
```

```python
for city in data.keys():
    print(city, "의 날씨는", data[city]["날씨상태"], "이고 온도는", data[city]["온도"], "입니다.")
```
