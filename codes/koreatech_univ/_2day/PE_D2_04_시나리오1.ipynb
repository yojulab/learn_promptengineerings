```python
# from google.colab import userdata
import os
from dotenv import load_dotenv

# .env 파일 로드
load_dotenv()

# 환경 변수에서 가져오기
api_key = os.getenv("OPENAI_API_KEY")

# Colab에 저장한 Secret에서 API 키 가져와 환경 변수에 등록
os.environ["OPENAI_API_KEY"] = api_key

# Colab에 저장한 Secret에서 API 키 가져와 환경 변수에 등록
# os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')

import openai

client = openai.OpenAI()
```

```python
import time
from pprint import pprint

roles_models = [
    ("당신은 친절한 초등학교 과학선생님입니다.", "gpt-4o-mini"),
    ("당신은 물리학과 대학교수입니다.", "gpt-4o-mini"),
    ("당신은 친절한 초등학교 과학선생님입니다.", "gpt-3.5-turbo")
]
prompt = "양자이론이 뭐에요?"
responses = []

for role, model in roles_models:
    response = client.chat.completions.create(
        messages= [
            {"role": "system", "content": role},
            {"role": "user", "content": prompt}
        ],
        model=model
    )
    responses.append(response.choices[0].message.content)
    time.sleep(1)

# 응답 내용 추출
pprint(responses)
```

```python
import pandas as pd

responses_df = pd.DataFrame(responses, columns=["response"])
responses_df
```

```python
prompt_eval = f"""
응답 텍스트에 대한 평가(1~5, 5점 만점)를 평가항목에 따라 예시와 같이 JSON형식으로 만들어주세요.

평가항목:
- 정확성: 응답 내용의 사실적 정확도와 일관성
- 일관성: 지정된 역할이나 톤을 응답이 꾸준히 유지하는 정도
- 톤 적절성: 목표 사용자나 상황에 맞는 어조/스타일 사용 여부
- 포맷 적합성: 요구된 출력 형식과 논리 구조를 제대로 따른 정도

예시:
[
    {{"정확성": 5, "일관성": 5, "톤 적절성": 5, "포맷 적합성": 5 }},
    {{"정확성": 4, "일관성": 4, "톤 적절성": 4, "포맷 적합성": 4 }},
    {{"정확성": 3, "일관성": 3, "톤 적절성": 3, "포맷 적합성": 3 }}
]

응답 텍스트:
{responses_df.to_markdown()} # LLM이 입력과 출력에 markdown에 대한 학습이 잘되어 있어 변환하는 것이 좋다.
"""

print(prompt_eval)
```

```python
# responses_df.to_markdown()
responses_df.to_csv("responses.csv", index=False)
```

```python
response = client.chat.completions.create(
    messages=[
        {"role": "user", "content": prompt_eval}
    ],
    model="gpt-4o",
    response_format={"type": "json_object"}
)
evaluation = response.choices[0].message.content
print(evaluation)
print(type(evaluation))
```

```python
import json

evals = json.loads(evaluation) # 문자열을 Dictionary로 변환

print(evals)

all_data = []
for scores in evals["평가"]: # Changed 'responses' to '평가'
    all_data.append({**scores})

eval_df = pd.DataFrame(all_data)
eval_df
```

```python
roles_models_df = pd.DataFrame(roles_models, columns=["role", "model"])
roles_models_df
```

```python
result_df = pd.concat([roles_models_df, responses_df, eval_df], axis=1)
result_df
```

```python
result_df.to_csv("result_evaluation.csv", index=False)
```
